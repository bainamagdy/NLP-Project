# TEXT PREPROCESSING

```python
data.drop(columns=["email_id","sender_email","has_attachment","num_characters","num_exclamation_marks"], inplace=True)
```

```python


nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))
```

```python
data['full_text'] = data['subject'].fillna('') + " " + data['email_text'].fillna('')

# عرض أول 5 صفوف للتأكد من الدمج
print(data[['subject', 'email_text', 'full_text']].head())
```

```python

# تأكد من تحميل هذه الحزم مرة واحدة على الأقل في جهازك



def get_wordnet_pos(treebank_tag):
    """
    دالة مساعدة لتحويل تصنيفات NLTK إلى تصنيفات يفهمها WordNet
    """
    if treebank_tag.startswith('J'):
        return wordnet.ADJ   # صفة (Adjective)
    elif treebank_tag.startswith('V'):
        return wordnet.VERB  # فعل (Verb)
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN  # اسم (Noun)
    elif treebank_tag.startswith('R'):
        return wordnet.ADV   # حال (Adverb)
    else:
        # إذا لم يتم التعرف عليه، افترض أنه اسم كقيمة افتراضية
        return wordnet.NOUN 

def preprocess_email_text(text):
    text = text.lower()
    
    # إزالة الرموز وعلامات الترقيم
    text = re.sub(r'[^\w\s]', ' ', text)
    
    # التقطيع
    words = text.split()
    
    # خطوة جديدة: استخراج نوع كل كلمة (POS Tagging)
    # تعيد قائمة من الأزواج، مثلاً: [('running', 'VBG'), ('fast', 'RB')]
    pos_tags = nltk.pos_tag(words)
    
    cleaned_words = []
    
    # نمر على الكلمة وتصنيفها معاً
    for word, tag in pos_tags:
        if word not in stop_words:
            # 1. تحويل التصنيف
            wordnet_pos = get_wordnet_pos(tag)
            
            # 2. إرجاع الكلمة لأصلها بناءً على نوعها المكتشف
            root_word = lemmatizer.lemmatize(word, pos=wordnet_pos)
            
            cleaned_words.append(root_word)
            
    return ' '.join(cleaned_words)
```

```python
print("جاري معالجة النصوص، يرجى الانتظار...")

# تطبيق الدالة لإنشاء عمود النص a
data['clean_text'] = data['full_text'].apply(preprocess_email_text)

# مقارنة سريعة بين النص الأصلي والنص بعد التنظيف
print(data[['full_text', 'clean_text']].head())
```

```python
# حذف الأعمدة النصية القديمة التي لم نعد بحاجة إليها
data = data.drop(columns=['subject', 'email_text', 'full_text'])

# التأكد من شكل الداتا النهائي
print("تمت المعالجة! الأعمدة الحالية في الداتا:")
print(data.info())
```

# FRATURE ENGINEERING 

```python
X_text = data['clean_text']

# نحتفظ بكل الأعمدة الرقمية ونحذف (النصوص، التصنيف، وأي عمود نصي مثل sender_domain)
X_numeric = data.drop(columns=['clean_text', 'label', 'sender_domain'])
y = data['label']

# ==========================================
# 2. تقسيم البيانات (Train/Test Split)
# ==========================================
# نقسم النصوص والأرقام معاً في نفس الوقت للحفاظ على الترتيب
X_text_train, X_text_test, X_num_train, X_num_test, y_train, y_test = train_test_split(
    X_text, X_numeric, y, test_size=0.2, random_state=42
)
```

```python
tfidf = TfidfVectorizer(max_features=5000, ngram_range=(1, 2))
X_text_train_tfidf = tfidf.fit_transform(X_text_train)
X_text_test_tfidf = tfidf.transform(X_text_test)
```

```python
scaler = StandardScaler()
X_num_train_scaled = scaler.fit_transform(X_num_train)
X_num_test_scaled = scaler.transform(X_num_test)
```

```python
X_train_combined = hstack((X_text_train_tfidf, X_num_train_scaled))
X_test_combined = hstack((X_text_test_tfidf, X_num_test_scaled))

print(f"شكل بيانات التدريب النهائية (نصوص + أرقام): {X_train_combined.shape}")
print(f"شكل بيانات الاختبار النهائية (نصوص + أرقام): {X_test_combined.shape}")
```

# MACHINE LEARNING MODELS

```python
print("--- نتائج نموذج Linear SVM ---")
svm_model = LinearSVC(random_state=42)
svm_model.fit(X_train_combined, y_train) # تدريب على البيانات المدمجة
svm_predictions = svm_model.predict(X_test_combined)
print(classification_report(y_test, svm_predictions))
```

```python
print("--- نتائج نموذج XGBoost ---")
xgb_model = XGBClassifier(
    n_estimators=100, 
    learning_rate=0.1, 
    random_state=42, 
    use_label_encoder=False, 
    eval_metric='logloss'
)
xgb_model.fit(X_train_combined, y_train) # تدريب على البيانات المدمجة
xgb_predictions = xgb_model.predict(X_test_combined)
print(classification_report(y_test, xgb_predictions))
```
